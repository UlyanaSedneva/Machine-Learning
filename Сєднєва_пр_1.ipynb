```python
# Підключення бібліотеки NumPy та генератора випадкових чисел
import numpy as np

rng = np.random.default_rng(19)   # фіксуємо seed, щоб результати відтворювались
np.set_printoptions(precision=4, suppress=True)
print("NumPy", np.__version__)
```

```python
# Завдання 1 (варіант 19).
# Створіть одновимірний масив (вектор) з добутками парних чисел на їхні попередні
# непарні числа, починаючи з 2, та виведіть його значення.
even = np.arange(2, 21, 2)        # парні числа 2, 4, ..., 20
odd = even - 1                    # попередні непарні числа 1, 3, ..., 19
products = even * odd             # 2*1, 4*3, 6*5, ...
print("Парні числа:     ", even)
print("Попередні непарні:", odd)
print("Добутки:         ", products)
```

```python
# Завдання 2 (варіант 19).
# Створіть двовимірний масив (матрицю) розміром 3x3, заповніть його значеннями
# квадратів випадкових цілих чисел від 1 до 100.
random_ints = rng.integers(1, 101, size=(3, 3))   # верхня межа не включається, тому 101
squares = random_ints ** 2
print("Випадкові цілі числа від 1 до 100:\n", random_ints)
print("Матриця їх квадратів:\n", squares)
```

```python
# Завдання 3 (варіант 19).
# Створіть масив розміром 5x5, заповніть його випадковими дійсними числами від -2 до 2
# та знайдіть середнє за модулем значення його елементів.
A3 = rng.uniform(-2, 2, size=(5, 5))
mean_abs = np.mean(np.abs(A3))
print("Масив 5x5:\n", A3)
print(f"Середнє за модулем значення елементів: {mean_abs:.4f}")
```

```python
# Завдання 4 (варіант 19).
# Створіть масив розміром 5x5, заповніть його випадковими дійсними числами в діапазоні
# від 0 до 1 та обчисліть суму елементів, які менше середнього значення.
A4 = rng.random(size=(5, 5))
mean4 = A4.mean()
below = A4[A4 < mean4]
print("Масив 5x5:\n", A4)
print(f"Середнє значення: {mean4:.4f}")
print(f"Елементів, менших за середнє: {below.size}")
print(f"Сума елементів, менших за середнє: {below.sum():.4f}")
```

```python
# Завдання 5 (варіант 19).
# Створіть два одновимірних масиви розміром 7, заповніть їх випадковими дійсними числами
# від -2 до 2 та виконайте на них поелементні операції додавання, віднімання та множення.
a5 = rng.uniform(-2, 2, size=7)
b5 = rng.uniform(-2, 2, size=7)
print("a     =", a5)
print("b     =", b5)
print("a + b =", a5 + b5)
print("a - b =", a5 - b5)
print("a * b =", a5 * b5)
```

```python
# Завдання 6 (варіант 19).
# Створіть дві матриці 4x4, заповніть їх довільними цілими числами від 7 до 17
# та знайдіть їх добуток.
A6 = rng.integers(7, 18, size=(4, 4))
B6 = rng.integers(7, 18, size=(4, 4))
C6 = A6 @ B6                      # матричний добуток (рядок на стовпчик)
print("A =\n", A6)
print("B =\n", B6)
print("A @ B =\n", C6)
```

```python
# Завдання 7 (варіант 19).
# Створіть дві матриці розміром 4x2 та 2x5, заповніть їх випадковими цілими числами
# в діапазоні від 1 до 6 та перемножте їх між собою.
A7 = rng.integers(1, 7, size=(4, 2))
B7 = rng.integers(1, 7, size=(2, 5))
C7 = np.dot(A7, B7)               # кількість стовпців A (2) = кількості рядків B (2)
print("A (4x2) =\n", A7)
print("B (2x5) =\n", B7)
print("A @ B (розмір", C7.shape, ") =\n", C7)
```

```python
# Завдання 8 (варіант 19).
# Створіть матрицю розміром 2x2, заповніть її випадковими цілими числами в діапазоні
# від 1 до 10 та знайдіть її обернену матрицю.
A8 = rng.integers(1, 11, size=(2, 2))
while np.isclose(np.linalg.det(A8), 0):     # обернена існує лише при det != 0
    A8 = rng.integers(1, 11, size=(2, 2))
A8_inv = np.linalg.inv(A8)
print("A =\n", A8)
print(f"det(A) = {np.linalg.det(A8):.4f}")
print("Обернена матриця A^-1 =\n", A8_inv)
print("Перевірка A @ A^-1 (має бути одинична):\n", A8 @ A8_inv)
```

```python
# Завдання 9 (варіант 19).
# Створіть матрицю розміром 3x3, заповніть її випадковими дійсними числами в діапазоні
# від 10 до 50 та транспонуйте її.
A9 = rng.uniform(10, 50, size=(3, 3))
print("A =\n", A9)
print("Транспонована A^T =\n", A9.T)
```

```python
# Завдання 10 (варіант 19).
# Створіть матрицю розміром 3x4 та вектор розміром 4x1, заповніть матрицю випадковими
# цілими числами в діапазоні від 1 до 10, а вектор - випадковими дійсними числами від 0 до 1,
# і знайдіть добуток матриці на вектор.
M10 = rng.integers(1, 11, size=(3, 4))
v10 = rng.random(size=(4, 1))
r10 = M10 @ v10                   # результат - вектор-стовпчик 3x1
print("Матриця 3x4 =\n", M10)
print("Вектор 4x1 =\n", v10)
print("Добуток (3x1) =\n", r10)
```

```python
# Завдання 11 (варіант 19).
# Створіть матрицю розміром 2x3 та вектор розміром 3x1, заповніть матрицю випадковими
# дійсними числами від 0 до 1, а вектор - випадковими цілими числами в діапазоні від 1 до 10,
# і знайдіть добуток матриці на вектор.
M11 = rng.random(size=(2, 3))
v11 = rng.integers(1, 11, size=(3, 1))
r11 = M11 @ v11                   # результат - вектор-стовпчик 2x1
print("Матриця 2x3 =\n", M11)
print("Вектор 3x1 =\n", v11)
print("Добуток (2x1) =\n", r11)
```

```python
# Завдання 12 (варіант 19).
# Створіть дві матриці розміром 3x3, заповніть їх випадковими цілими числами в діапазоні
# від 1 до 5 та знайдіть добуток елементів діагоналей обох матриць (окремо).
A12 = rng.integers(1, 6, size=(3, 3))
B12 = rng.integers(1, 6, size=(3, 3))
print("A =\n", A12)
print("Головна діагональ A:", np.diag(A12), "-> добуток =", np.prod(np.diag(A12)))
print("B =\n", B12)
print("Головна діагональ B:", np.diag(B12), "-> добуток =", np.prod(np.diag(B12)))
```

```python
# Завдання 13 (варіант 19).
# Створіть дві матриці розміром 3x3, заповніть їх випадковими дійсними числами в діапазоні
# від -1 до 1 та знайдіть суму їх елементів.
A13 = rng.uniform(-1, 1, size=(3, 3))
B13 = rng.uniform(-1, 1, size=(3, 3))
S13 = A13 + B13                   # поелементна сума матриць
print("A =\n", A13)
print("B =\n", B13)
print("Сума відповідних елементів A + B =\n", S13)
print(f"Сума всіх елементів обох матриць: {S13.sum():.4f}")
```

```python
# Завдання 14 (варіант 19).
# Створіть матрицю розміром 3x3, заповніть її випадковими дійсними числами в діапазоні
# від -3 до 3 та знайдіть найбільший елемент матриці.
A14 = rng.uniform(-3, 3, size=(3, 3))
max14 = A14.max()
row, col = np.unravel_index(A14.argmax(), A14.shape)
print("A =\n", A14)
print(f"Найбільший елемент: {max14:.4f} (рядок {row}, стовпець {col})")
```

```python
# Завдання 15 (варіант 19).
# Створіть матрицю розміром 3x3, заповніть її випадковими дійсними числами в діапазоні
# від -0.5 до 0.5 та знайдіть вектор-стовпчик, що містить кількість нульових елементів
# в кожному рядку матриці.
A15 = rng.uniform(-0.5, 0.5, size=(3, 3))
zeros_exact = np.count_nonzero(A15 == 0, axis=1).reshape(-1, 1)
print("A =\n", A15)
print("Кількість нульових елементів у кожному рядку (вектор-стовпчик):\n", zeros_exact)

# Ймовірність отримати рівно 0 серед неперервних випадкових чисел практично нульова,
# тому всі значення вектора дорівнюють 0. Для наочності додатково порахуємо
# «майже нульові» елементи з |x| < 0.1.
near_zero = np.count_nonzero(np.abs(A15) < 0.1, axis=1).reshape(-1, 1)
print("Кількість елементів з |x| < 0.1 у кожному рядку:\n", near_zero)
```
